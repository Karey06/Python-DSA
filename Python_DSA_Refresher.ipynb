```python
# Question 1: Days into seconds

days = int(input("Enter number of days: "))
total_seconds = days * 24 * 60 * 60

print("Total seconds:", total_seconds)
```

```python
# Question 2: Volume of a sphere

r = float(input("Enter the radius: "))
pi = 3.142
volume = (4 / 3) * pi * (r**3)

print("The volume of the sphere is", volume)

```

```python
# Question 3: Area and perimeter of a square using functions

def area(length):
    return length * length

def perimeter(length):
    return 4 * length

length = float(input("Enter length of square: "))

print("Area:", area(length))
print("Perimeter:", perimeter(length))

```

```python
# Question 4: Determining if a character input is uppercase or lowercase

def check_letter(char):
    if char >= "A" and char <= "Z":
        print("uppercase")
    elif char >= "a" and char <= "z":
        print("lowercase")
    else:
        print("not a letter")


char = input("Enter a letter: ")
check_letter(char)
```

```python
# Question 5: Python program equivalent of a given pseudocode

x = 0
y = 20

while True:
    y = y - 4
    x = x + 2 / y
    if y < 6:
        break

print("x is:", x)
```

```python
# Question 6: Entering 5 numbers and finding average

numbers = []

for i in range(5):
    num = float(input("Enter number: "))
    numbers.append(num)

total = 0
for n in numbers:
    total = total + n

average = total / 5

print("The numbers are:", numbers)
print("The average is:", average)
```
